Aritificial Neuron

```python
import numpy as np

def artificial_neuron(inputs, weights, bias=0.0, activation='step'):
    z = np.dot(inputs, weights) + bias
    if activation == 'step':
        return np.where(z >= 0, 1, 0)
    elif activation == 'sigmoid':
        return 1 / (1 + np.exp(-z))
    elif activation == 'relu':
        return np.maximum(0, z)
    else:
        raise ValueError("Invalid activation")

x_example = np.array([0.5, 0.3])
w_example = np.array([0.8, -0.2])
b = 5.0

print(f"Step Activation: {artificial_neuron(x_example, w_example, b, 'step')}")
print(f"Sigmoid Activation: {artificial_neuron(x_example, w_example, b, 'sigmoid')}")
print(f"ReLU Activation: {artificial_neuron(x_example, w_example, b, 'relu')}")
```
